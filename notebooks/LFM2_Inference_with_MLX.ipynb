# 💧 LFM2 Inference with MLX

This notebook demonstrates how to run [LFM2](https://huggingface.co/collections/LiquidAI/lfm2-67d775f3b4b6fe79fbb21bda) and [LFM2.5](https://huggingface.co/collections/LiquidAI/lfm25-6839e3e26b2a9fdbde95b341) models using [MLX](https://github.com/ml-explore/mlx), Apple's machine learning framework.

> ⚠️ **Note:** This notebook is for demonstration purposes. MLX is optimized for Apple Silicon (M1/M2/M3/M4) Macs with Metal GPU acceleration. For best performance, run this code locally on a Mac with an M-series chip.

## Basic Text Generation

Load the model and generate text:

```python
from mlx_lm import load, generate

# Load model and tokenizer
model, tokenizer = load("mlx-community/LFM2-1.2B-8bit")

# Apply chat template
prompt = "What is machine learning?"
messages = [{"role": "user", "content": prompt}]
formatted_prompt = tokenizer.apply_chat_template(
    messages, tokenize=False, add_generation_prompt=True
)

# Generate response
response = generate(model, tokenizer, prompt=formatted_prompt, verbose=True)
print(response)
```

## Generation Parameters

Control generation with custom parameters:

```python
prompt = "Explain quantum computing in simple terms."
messages = [{"role": "user", "content": prompt}]
formatted_prompt = tokenizer.apply_chat_template(
    messages, tokenize=False, add_generation_prompt=True
)

response = generate(
    model,
    tokenizer,
    prompt=formatted_prompt,
    temperature=0.3,
    top_p=0.9,
    max_tokens=512,
    repetition_penalty=1.05
)
print(response)
```

## Streaming Generation

Stream responses as they're generated:

```python
from mlx_lm import stream_generate

prompt = "Tell me a story about space exploration."
messages = [{"role": "user", "content": prompt}]
formatted_prompt = tokenizer.apply_chat_template(
    messages, tokenize=False, add_generation_prompt=True
)

for token in stream_generate(model, tokenizer, prompt=formatted_prompt, max_tokens=512):
    print(token, end="", flush=True)
```

## Vision Models

LFM2-VL models support image inputs using `mlx_vlm`:

```python
!pip install mlx-vlm pillow
```

```python
from mlx_vlm import load, generate
from mlx_vlm.prompt_utils import apply_chat_template
from PIL import Image
import requests
from io import BytesIO

# Load vision model
model, processor = load("mlx-community/LFM2-VL-1.6B-8bit")

# Load image from URL
image_url = "https://cdn.britannica.com/61/93061-050-99147DCE/Statue-of-Liberty-Island-New-York-Bay.jpg"
response = requests.get(image_url)
image = Image.open(BytesIO(response.content))

# Create prompt
messages = [
    {
        "role": "user",
        "content": [
            {"type": "image"},
            {"type": "text", "text": "What's in this image?"}
        ]
    }
]

# Apply chat template and generate
prompt = apply_chat_template(processor, messages)
output = generate(model, processor, image, prompt, verbose=False)
print(output)
```
